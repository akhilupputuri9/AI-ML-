Q1. Movie Ticket Counter

A cinema charges ticket prices based on age:

Below 5 years : Free
5 to 17 years : Rs. 100
18 to 59 years : Rs. 200
60 and above : Rs. 120

```python
age = int(input ("enter your age: "))
if age < 5:
    print ("Below 5Yrs children are free ")
elif 5<= age <= 17:
    print ("Ticket price is Rs.100")
elif 18<= age <= 59:
    print ("Ticket price is Rs.200")
else :
    print ("Ticket price is Rs.120")

```

Q2. Electricity Bill Calculator

--------------------------------------------------------------

The electricity board charges as follows:

First 100 units : Rs. 3 per unit
Next 100 units (101-200): Rs. 5 per unit
Above 200 units : Rs. 8 per unit


Take the number of units consumed and print the total bill.



Sample run:

Enter units consumed: 250

Total bill: Rs. 1200

(100*3 + 100*5 + 50*8 = 300 + 500 + 400)

```python
units = int(input("enter the total no.of units: "))
if units <100:
    bill = units*3
elif units <=200:
    bill = (100*3)+((units-100)*5)
else:
    bill = (100*3)+ (100*5)+ ((units-200)*8)
print ("Total bill:"bill")
```
